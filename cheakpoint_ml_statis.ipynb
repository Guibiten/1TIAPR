```python
# BIBLIOTECAS
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# PARTE 1: STATISTICAL COMPUTING (Exploração, Tratamento e Validação)
print("--- PARTE 1: EXPLORAÇÃO E TRATAMENTO DE DADOS ---\n")

# 1. Coleta dos dados
# Carregando o dataset IRIS original
iris = load_iris()

# Transformando os dados em um DataFrame (tabela) do Pandas para facilitar a análise
df = pd.DataFrame(iris.data, columns=iris.feature_names)
# Adicionando a coluna 'target' que contém a espécie da flor (0=Setosa, 1=Versicolor, 2=Virginica)
df['especie'] = iris.target

# 2. Exploração da estrutura e características dos dados
print(f"-> Tamanho original do dataset: {df.shape[0]} linhas e {df.shape[1]} colunas.")
print("\n-> Primeiras 5 linhas do dataset:")
display(df.head()) # Mostra as primeiras linhas de forma visual

print("\n-> Resumo estatístico (Média, Mínimo, Máximo):")
display(df.describe()) # Mostra as estatísticas básicas das medidas

# 3. Identificação e tratamento de valores ausentes e duplicidades
valores_ausentes = df.isnull().sum().sum()
linhas_duplicadas = df.duplicated().sum()

print(f"\n-> Valores ausentes (nulos) encontrados: {valores_ausentes}")
print(f"-> Linhas duplicadas encontradas: {linhas_duplicadas}")

# Tratamento: Se houver linhas duplicadas, o código irá removê-las
if linhas_duplicadas > 0:
    df = df.drop_duplicates()
    print("-> AÇÃO: Linhas duplicadas foram removidas com sucesso!")

# 4. Análise da qualidade dos dados (Validação)
print("\n-> CONCLUSÃO DA PARTE 1:")
print("Os dados foram explorados e validados. Não há valores ausentes (nulos) e as duplicidades foram tratadas.")
print("O dataset está limpo, estruturado e pronto para a aplicação de Machine Learning.\n")
print("=" * 70 + "\n")


# PARTE 2: MACHINE LEARNING (Modelo KNN)
print("--- PARTE 2: TREINAMENTO DO MODELO KNN ---\n")

# 1. Preparação dos dados
# Separando as variáveis preditoras (X - características) da variável alvo (y - espécie)
X = df.drop('especie', axis=1) # X recebe todas as colunas, exceto a 'especie'
y = df['especie']              # y recebe apenas a coluna 'especie' (o gabarito)

# 2. Separação dos conjuntos de treino e teste
# Dividindo: 70% para treino (aprender) e 30% para teste (avaliar)
# random_state=42 garante que o sorteio dos dados seja sempre o mesmo
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Treinamento do modelo testando diferentes valores de k
valores_k = [1, 3, 5, 7, 9] # Valores exigidos no exercício
resultados = []             # Lista vazia para guardar a pontuação de cada K

print("Treinando os modelos e calculando a acurácia...\n")

# Loop para testar automaticamente cada valor de K
for k in valores_k:
    # Instancia o modelo KNN com o número de vizinhos atual
    knn = KNeighborsClassifier(n_neighbors=k)

    # Treina o modelo usando os dados de treino
    knn.fit(X_train, y_train)

    # Realiza a predição (prova) usando os dados de teste
    y_pred = knn.predict(X_test)

    # Calcula a métrica de acurácia (taxa de acertos)
    acuracia = accuracy_score(y_test, y_pred)

    # Salva o resultado na lista
    resultados.append({'Valor de K (Vizinhos)': k, 'Acurácia': acuracia})

# 4. Apresentação dos resultados em tabela comparativa
print("-> TABELA COMPARATIVA DE RESULTADOS:")
# Converte a lista de resultados numa tabela elegante do Pandas
tabela_resultados = pd.DataFrame(resultados)

# Exibe a tabela na tela
display(tabela_resultados)
```
